```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import scipy.stats as stats
import os
import warnings
warnings.filterwarnings("ignore")

sns.set_style("whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)
plt.rcParams["axes.titlesize"] = 13
plt.rcParams["axes.labelsize"] = 11

# Upload train.csv if it is not already in the Colab session
if not os.path.exists("train.csv"):
    from google.colab import files
    print("Please upload train.csv")
    files.upload()

df = pd.read_csv("train.csv")
print("Loaded OK")
```

```python
print("Shape (rows, columns):", df.shape)
print()
print("First 5 rows:")
display(df.head())
print()
df.info()
```

```python
missing = df.isnull().sum().sort_values(ascending=False)
missing = missing[missing > 0]
missing_table = pd.DataFrame({
    "missing_count": missing,
    "missing_percent": (missing / len(df) * 100).round(2)
})
print("Columns with missing values:", len(missing_table))
display(missing_table)

# Keep an untouched copy for reference
df_raw = df.copy()
```

```python
numeric_cols = df.select_dtypes(include="number").columns.tolist()
categorical_cols = df.select_dtypes(include="object").columns.tolist()

ordinal_map = {
    "ExterQual":   ["Po", "Fa", "TA", "Gd", "Ex"],
    "KitchenQual": ["Po", "Fa", "TA", "Gd", "Ex"],
}

for col, order in ordinal_map.items():
    df[col] = pd.Categorical(df[col], categories=order, ordered=True)

print("Numeric columns:", len(numeric_cols))
print("Categorical (object) columns:", len(categorical_cols))
print("Ordinal columns converted:", list(ordinal_map.keys()))

# Summary table: column | type | unique values | examples
def col_type(c):
    if c in ordinal_map or c == "OverallQual" or c == "OverallCond":
        return "ordinal"
    if c in categorical_cols:
        return "categorical (nominal)"
    if df[c].nunique() <= 20:
        return "numerical (discrete)"
    return "numerical (continuous)"

summary = pd.DataFrame({
    "column": df.columns,
    "type": [col_type(c) for c in df.columns],
    "n_unique": [df[c].nunique() for c in df.columns],
    "examples": [", ".join(map(str, df[c].dropna().unique()[:3])) for c in df.columns],
})
display(summary)
print()
print(summary["type"].value_counts())
```

```python
for col in ["SalePrice", "GrLivArea", "LotArea"]:
    s = df[col]
    mean, median, mode = s.mean(), s.median(), s.mode()[0]
    std, var = s.std(), s.var()
    rng = s.max() - s.min()
    iqr = s.quantile(0.75) - s.quantile(0.25)
    skew, kurt = stats.skew(s), stats.kurtosis(s)

    print(f"===== {col} =====")
    print(f"mean={mean:,.2f} | median={median:,.2f} | mode={mode:,.2f}")
    print(f"std={std:,.2f} | variance={var:,.2f} | range={rng:,.2f} | IQR={iqr:,.2f}")
    print(f"skewness={skew:.3f} | kurtosis (excess)={kurt:.3f}")
    if mean > median:
        print("NOTE: mean > median -> right skewed")
    elif mean < median:
        print("NOTE: mean < median -> left skewed")
    else:
        print("NOTE: mean = median -> roughly symmetric")
    print()

for col in ["Neighborhood", "HouseStyle"]:
    print(f"===== {col} =====")
    print(df[col].value_counts())
    print("Mode:", df[col].mode()[0])
    print()
```

```python
# 1. Duplicates
dup_count = df.duplicated().sum()
print("Duplicate rows:", dup_count)
if dup_count > 0:
    df = df.drop_duplicates()
    print("New shape:", df.shape)

# 2. Inconsistent categories (casing, extra spaces)
problems = {}
for col in categorical_cols:
    if df[col].dtype != "object":
        continue
    vals = df[col].dropna().unique()
    cleaned = {v.strip().lower() for v in vals}
    if len(cleaned) != len(vals) or any(v != v.strip() for v in vals):
        problems[col] = list(vals)

if problems:
    print("Columns with inconsistent labels:", list(problems.keys()))
    for col in problems:
        df[col] = df[col].str.strip()
else:
    print("No casing / extra-space problems found in categorical columns.")
```

```python
# NA = feature absent (dataset documentation)
none_cols = [
    "PoolQC", "MiscFeature", "Alley", "Fence", "FireplaceQu",
    "GarageType", "GarageFinish", "GarageQual", "GarageCond",
    "BsmtQual", "BsmtCond", "BsmtExposure", "BsmtFinType1", "BsmtFinType2",
]
for col in none_cols:
    df[col] = df[col].fillna("None")

# Numeric: LotFrontage -> median grouped by Neighborhood
df["LotFrontage"] = df.groupby("Neighborhood")["LotFrontage"].transform(lambda s: s.fillna(s.median()))
df["LotFrontage"] = df["LotFrontage"].fillna(df["LotFrontage"].median())  # safety net

# MasVnrArea: NA = no masonry veneer -> 0 and type "None"
df["MasVnrArea"] = df["MasVnrArea"].fillna(0)
df["MasVnrType"] = df["MasVnrType"].fillna("None")

# GarageYrBlt: NA when there is no garage -> use YearBuilt as a neutral placeholder
df["GarageYrBlt"] = df["GarageYrBlt"].fillna(df["YearBuilt"])

# Electrical: 1 genuinely missing value -> mode
df["Electrical"] = df["Electrical"].fillna(df["Electrical"].mode()[0])

remaining = df.isnull().sum().sum()
print("Remaining missing values:", remaining)
assert remaining == 0, "Some columns still have missing values"
```

```python
# Impossible values
print("YearBuilt > YrSold:", (df["YearBuilt"] > df["YrSold"]).sum())
print("YearRemodAdd < YearBuilt:", (df["YearRemodAdd"] < df["YearBuilt"]).sum())
for col in ["LotArea", "GrLivArea", "TotalBsmtSF", "GarageArea", "SalePrice"]:
    print(f"Negative values in {col}:", (df[col] < 0).sum())
```

```python
num_df = df.select_dtypes(include="number")

cov_matrix = num_df.cov()
corr_matrix = num_df.corr(method="pearson")

top_corr = corr_matrix["SalePrice"].abs().sort_values(ascending=False)
top10 = top_corr.index[1:11].tolist()  # skip SalePrice itself

print("Top 10 features correlated with SalePrice:")
print(corr_matrix.loc[top10, "SalePrice"].round(3))

print()
for feat in top10[:3]:
    print(f"{feat}: covariance with SalePrice = {cov_matrix.loc[feat, 'SalePrice']:,.1f} | "
          f"correlation = {corr_matrix.loc[feat, 'SalePrice']:.3f}")
```

```python
def iqr_outliers(series):
    q1, q3 = series.quantile(0.25), series.quantile(0.75)
    iqr = q3 - q1
    lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    return df[(series < lower) | (series > upper)]

def zscore_outliers(series, threshold=3):
    z = (series - series.mean()) / series.std()
    return df[z.abs() > threshold]

results = {}
for col in ["GrLivArea", "SalePrice"]:
    a = iqr_outliers(df[col])
    b = zscore_outliers(df[col])
    overlap = a.index.intersection(b.index)
    results[col] = (a, b)
    print(f"=== {col} ===")
    print(f"IQR outliers:     {len(a)}")
    print(f"Z-score outliers: {len(b)}")
    print(f"Overlap:          {len(overlap)}")
    print()
```

```python
# DECISION: remove the very large, very cheap, non-typical sales; keep all other outliers
bad = (df["GrLivArea"] > 4000) & (df["SalePrice"] < 300000)
print("Rows removed as non-typical sales:", bad.sum())
display(df.loc[bad, ["Id", "GrLivArea", "SalePrice", "SaleCondition", "Neighborhood"]])

df_clean = df[~bad].copy()
print("Shape before:", df.shape, "| after:", df_clean.shape)

print()
print("Correlation GrLivArea-SalePrice before removal:", round(df["GrLivArea"].corr(df["SalePrice"]), 3))
print("Correlation GrLivArea-SalePrice after removal: ", round(df_clean["GrLivArea"].corr(df_clean["SalePrice"]), 3))
```
